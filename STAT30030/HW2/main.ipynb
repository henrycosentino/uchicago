## (i)
- Generate 500 samples of 2 random variables X1 and X2 with distribution Exp(1). For each of the 500 samples, calculate min(X1,X2) max(X1,X2), R = max(X1,X2)− min(X1,X2), and (X1− X2). This should give you 500 tuples/arrays, each of them constructed using 2 independent draws. Create a scatter plot of X1− X2 versus min(X1,X2) and comment on if it corresponds to what you proved in Problem 7.

```python
n: int = 500                    # Sample size
lambda_: float = 1.0            # lambda parameter for exponential distribution
dist = np.random.exponential    # Exponential distribution object
```

```python
samples = []

for _ in range(0, n):
    x1 = dist(scale = 1 / lambda_)  # Draw 1
    x2 = dist(scale = 1 / lambda_)  # Draw 2

    max_ = max(x2, x1)
    min_ = min(x2, x1)

    R = max_ - min_

    X1_X2 = x1 - x2

    samples.append((x1, x2, max_, min_, R, X1_X2))
```

```python
df = pd.DataFrame(
    data = samples,
    columns = ['X1', 'X2', 'max(X1,X2)', 'min(X1,X2)', 'R', 'X1-X2']
)

df.head(2)
```

```python
x = df['X1-X2']
y = df['min(X1,X2)']

fig = px.scatter(
    x = x,
    y = y,
    labels = {'x': 'X1 - X2', 'y': 'min(X1, X2)'},
    title = 'X1 - X2 vs. min(X1, X2)',
    subtitle= f'Corr(X1 - X2, min(X1, X2)) = {np.corrcoef(x, y)[1,0]:0.4f}',
)
fig.show()
```

### Comment:


Yes, the scatter plot corresponds to what I proved. The scatter plot shows a visible relationship between $X_1 - X_2$ and $\min(X_1, X_2)$.

## (ii)

- Create two scatterplots to investigate (visually) if: 
    - (a) X1 is independent of min(X1,X2); 
    - (b) max(X1,X2) is independent of X1− X2. 
- Comment on how the scatterplots can inform you on ideas for a proof.

```python
x = df['X1']
y = df['min(X1,X2)']

fig = px.scatter(
    x = x,
    y = y,
    title = 'Dependency Structure Exploration: X1 vs. min(X1, X2)',
    subtitle= f'Corr(X1, min(X1, X2)) = {np.corrcoef(x, y)[1,0]:0.4f}',
    labels = {'x': 'X1', 'y': 'min(X1, X2)'},
)
fig.show()
```

### Comment:

The scatter plot shows a linear relationship between $X_1$ and $\min(X_1, X_2)$, so these random variables are dependent (corroborated by a correlation of 0.4340). To prove dependency, we just have to show that one is necessarily a function of the other. In this case, I would use the pdf for $\min(X_1, X_2)$ and show it is dependent on the value $X_1$ takes.

```python
x = df['X1-X2']
y = df['max(X1,X2)']

fig = px.scatter(
    x = x,
    y = y,
    title = 'Dependency Structure Exploration: X1 - X2 vs. max(X1, X2)',
    subtitle= f'Corr(X1 - X2, max(X1, X2)) = {np.corrcoef(x, y)[1,0]:0.4f}',
    labels = {'x': 'X1 - X2', 'y': 'max(X1, X2)'},
)
fig.show()
```

### Comment:

The scatter plot shows a non-linear relationship between $X_1 - X_2$ and $\min(X_1, X_2)$, so these random variables are dependent. I have included the correlation, which measures linear relationships, but since the dependency structure is non-linear, the correlation acts as a poor informant of dependence. The non-linear relationship signals that I may have to use cases in my proof: $X_1 > X_2$ and $X_1 < X_2$.
