```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

## 12 (a)

```python
def sample_and_plot(n: int = 20, seed: None | int = 21):
    if seed:
        rng = np.random.default_rng(seed=seed)
        X = rng.normal(-1, 1, size=n)
    else:
        X = np.random.normal(-1, 1, size=n)
    
    X_plus = np.maximum(0, X)
    
    grid = np.linspace(-2, 3, 500)

    true_cdf = np.where(grid < 0, 0, stats.norm.cdf(grid, loc=-1, scale=1))
    empirical_cdf = np.mean(X_plus[:, None] <= grid, axis=0)

    plt.figure(figsize=(8, 5))
    plt.plot(grid, true_cdf, label="True CDF", color="black", linewidth=1.5)
    plt.step(grid, empirical_cdf, label=f"Empirical CDF", color="orange", linewidth=1.8)

    plt.ylabel('F(x)')
    plt.xlabel('x')
    plt.title(f"Empirical vs. Theoretical CDF for n = {n}", fontweight='bold')

    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
sample_and_plot(n = 20, seed = 21)
```

```python
sample_and_plot(n = 50, seed = 21)
```

```python
sample_and_plot(n = 100, seed = 21)
```

## 12 (b)

```python
n = 1_000

rng = np.random.default_rng(seed = 42)

X = rng.normal(0, 1, size=n)
Y = np.sin(X)

Y[:3]
```

```python
y_sorted = np.sort(Y)
cdf_values = np.arange(1, n + 1) / n
```

```python
plt.figure(figsize=(10, 5))

plt.plot(y_sorted, cdf_values)

plt.title(f"Empirical CDF of Y = sin(X) for n = {n}", fontweight='bold')
plt.xlabel('y')
plt.ylabel('F(y)')

plt.tight_layout()
plt.show()
```
